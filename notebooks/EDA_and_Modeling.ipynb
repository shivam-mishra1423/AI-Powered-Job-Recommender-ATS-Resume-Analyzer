# AI Job Recommender — EDA & Modeling
End-to-end walkthrough: jobs dataset EDA, TF-IDF index, recommendations, ATS scoring.

```python
import pandas as pd, numpy as np, matplotlib.pyplot as plt, seaborn as sns
df = pd.read_csv('../data/jobs.csv')
print(df.shape); df.head()
```

```python
df['title'].value_counts().head(15).plot(kind='barh', figsize=(8,6))
plt.title('Top job titles'); plt.gca().invert_yaxis(); plt.show()
```

```python
df['location'].value_counts().head(15).plot(kind='bar', figsize=(10,4))
plt.title('Top locations'); plt.xticks(rotation=45, ha='right'); plt.show()
```

```python
from collections import Counter
c = Counter()
for s in df['skills'].dropna(): c.update([x.strip() for x in s.split(',')])
pd.Series(dict(c.most_common(20))).plot(kind='barh', figsize=(8,6))
plt.title('Top in-demand skills'); plt.gca().invert_yaxis(); plt.show()
```
